# **My First ML Project**

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
df
```

```python
y = df['logS']
y
```

```python
x = df.drop('logS', axis=1)
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
y_lr_train_pred = lr.predict(x_train)
y_lr_test_pred = lr.predict(x_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test_pred
```

```python
from sklearn.metrics import mean_squared_error
lr_train_mse = mean_squared_error(y_train, y_lr_train_pred)
lr_test_mse = mean_squared_error(y_test, y_lr_test_pred)
```

```python
print('LR MSE (Train): ', lr_train_mse)
print('LR MSE (Test): ', lr_test_mse)
```

```python
lr_results = pd.DataFrame(['Linear Regression', lr_train_mse, lr_test_mse]).transpose()
lr_results.columns = ['Method', 'Training MSE', 'Test MSE']
lr_results
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(max_depth=2, random_state=100)
rf.fit(x_train, y_train)
```

```python
y_rf_train_pred = rf.predict(x_train)
y_rf_test_pred = rf.predict(x_test)
```

```python
from  sklearn.metrics import mean_squared_error
rf_train_mse = mean_squared_error(y_train, y_rf_train_pred)
rf_test_mse = mean_squared_error(y_test, y_rf_test_pred)
```

```python
rf_results = pd.DataFrame(['Random Forest', rf_train_mse, rf_test_mse]).transpose()
rf_results.columns = ['Method', 'Training MSE', 'Test MSE']
rf_results
```

```python
df_results = pd.concat([lr_results, rf_results], axis=0)
df_results.columns = ['Method', 'Training MSE', 'Test MSE']
df_results
```
